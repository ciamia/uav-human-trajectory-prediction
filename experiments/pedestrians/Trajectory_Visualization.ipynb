```python
# ================================
# TRAJECTRON++ ABSOLUTE IMPORT FIX (FINAL)
# ================================
import sys
from pathlib import Path
import types
import importlib

# 🔹 Proje root bul
ROOT = Path.cwd()
while not (ROOT / "trajectron").exists():
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT))

# ================================
# 1️⃣ ÖNCE BOŞ model MODÜLÜ OLUŞTUR
# ================================
model = types.ModuleType("model")
sys.modules["model"] = model

# ================================
# 2️⃣ ŞİMDİ trajectron.model ALT MODÜLLERİNİ BAĞLA
# ================================
tmodel = importlib.import_module("trajectron.model")
tmodel_dataset = importlib.import_module("trajectron.model.dataset")

model.mgcvae = importlib.import_module("trajectron.model.mgcvae")
model.dataset = tmodel_dataset

sys.modules["model.dataset"] = tmodel_dataset
sys.modules["model.mgcvae"] = model.mgcvae

# ================================
# 3️⃣ ENVIRONMENT alias
# ================================
environment = types.ModuleType("environment")
sys.modules["environment"] = environment

ds = importlib.import_module("trajectron.environment.data_structures")
environment.RingBuffer = ds.RingBuffer
environment.SingleHeaderNumpyArray = ds.SingleHeaderNumpyArray
environment.DoubleHeaderNumpyArray = ds.DoubleHeaderNumpyArray

tenv = importlib.import_module("trajectron.environment")
for attr in dir(tenv):
    if not attr.startswith("__"):
        setattr(environment, attr, getattr(tenv, attr))

print("✅ model + environment imports FIXED")
```
